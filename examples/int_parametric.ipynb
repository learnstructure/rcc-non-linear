#AL-Hawarneh and Alam 

```python
from rcc_non_linear import (
    Model,
    HingeRadau,
    HingeMidpoint,
    Lobatto,
    Legendre,
    NewtonCotes,
    Trapezoidal,
    CompositeSimpson,
)
```

```python
base_props = {
    'fc': 4.484,
    'D': 12, 'L': 68.11, 'cover': 0.984,
    'nBars': 14, 'db': 0.4449,
    'fy': 60.191, 'fu': 108, 'e_ult': 0.1,
    'dh':0.325, 'sh':2.756, 'fyh':60.191, 'esm':0.1,
    'P_axial': 45.2
}
```

```python
integrations = {
    "HingeRadau (default)": HingeRadau(),
    # "HingeRadau (custom lp=20)": HingeRadau(lp_bottom=20.0),
    # "HingeMidpoint": HingeMidpoint(),
    "Lobatto (5 points)": Lobatto(num_points=5),
    "Legendre (5 points)": Legendre(num_points=5),
    # "NewtonCotes (5 points)": NewtonCotes(num_points=5),
}

for name, integration_obj in integrations.items():
    props = base_props.copy()
    props['integration'] = integration_obj
    
    print(f"\n--- Testing Integration: {name} ---")
    model = Model(props)
    results_df, bilinear_df, yield_step = model.run_pushover_analysis(dU=0.01)
    peak_force = max(results_df['forces'])
    max_disp = max(results_df['displacements'])
    print(f"Max Drift: {(max_disp/model.L*100):.2f}%, Peak Lateral Force: {peak_force:.2f} kips, Yield Step: {yield_step}")
```
